## Export the data

### Microsoft data

#### Drop `Adj Close`

- To not confuse the `Close` and `Adj Close`

```python
df_msft = df_msft.drop(columns='Adj Close')
```

#### To Excel

- Create a folder called `data`

```python
 df_msft.to_excel('data/microsoft_stck_price.xlsx')
```

#### To CSV

- Less size in memory

```python
df_msft.to_csv('data/microsoft_stock_price.csv')
```

### Bitcoin data altogether (step by step)

#### Define ticker

```python
ticker = 'BTC-USD'
ticker
```

#### Download OLHVC

```python
df_ticker = yf.download(ticker, multi_level_index=False, auto_adjust=False)
df_ticker
```
